```python
# 2. Passar o texto para minúsculas

nlp = spacy.load("en_core_web_sm")  

def smart_lower_spacy(text):
    doc = nlp(text)
    result = []
    for token in doc:

        if token.ent_type_:
            result.append(token.text_with_ws)
        else:
            result.append(token.lower_ + token.whitespace_)
    return "".join(result)


df_test["text"] = df_test["text"].apply(smart_lower_spacy)
```

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
from nltk.corpus import wordnet

def get_wordnet_pos(word):
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {
        "J": wordnet.ADJ,
        "N": wordnet.NOUN,
        "V": wordnet.VERB,
        "R": wordnet.ADV
    }
    return tag_dict.get(tag, wordnet.NOUN)

def lemmatize_text(text):
    tokens = text.split()
    lemmatized = [
        lemmatizer.lemmatize(word, get_wordnet_pos(word))
        for word in tokens
    ]
    return ' '.join(lemmatized)
```

5.1 Aplicação dos Modelos baseados em Léxicos e regras

```python
# 5.A Previsão usando TextBlob
from textblob import TextBlob

def textblob_predict(text):
    polarity = TextBlob(text).sentiment.polarity
    return 1 if polarity >= 0 else 0

df_models["pred_textblob"] = df_models["text"].apply(textblob_predict)

print(df_models[["text", "label", "pred_textblob"]].head())
```

```python
# 5.B Previsão usando VADER
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

vader = SentimentIntensityAnalyzer()

def vader_predict(text):
    score = vader.polarity_scores(text)["compound"]
    return 1 if score >= 0 else 0

df_models["pred_vader"] = df_models["text"].apply(vader_predict)

print(df_models[["text", "label", "pred_vader"]].head())
```

```python
import stanza
from stanza.models.common.doc import Document
from tqdm import tqdm

stanza_pipeline = stanza.Pipeline(
    lang="en",
    processors="tokenize,sentiment",
    use_gpu=False
)

def stanza_predict_batch(texts, batch_size=200):
    preds = []

    for i in tqdm(range(0, len(texts), batch_size)):
        batch = texts.iloc[i:i+batch_size].tolist()

        # converter strings em Document objects
        in_docs = [Document([], text=t) for t in batch]

        # processar o lote
        out_docs = stanza_pipeline(in_docs)

        for doc in out_docs:
            sentiments = [sentence.sentiment for sentence in doc.sentences]

            if len(sentiments) == 0:
                preds.append(1)
            else:
                avg_sentiment = sum(sentiments) / len(sentiments)
                # 0=negative, 1=neutral, 2=positive
                preds.append(1 if avg_sentiment >= 1 else 0)

    return preds

df_models["pred_stanza"] = stanza_predict_batch(df_models["text"], batch_size=200)

print(df_models[["text", "label", "pred_stanza"]].head())
```

```python
# 5.D Previsão usando SentiWordNet
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("sentiwordnet")
nltk.download("wordnet")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")

from nltk.corpus import sentiwordnet as swn
from nltk.tokenize import word_tokenize

def sentiwordnet_predict(text):
    tokens = word_tokenize(text)
    tagged = nltk.pos_tag(tokens)
    
    sentiment_score = 0
    valid_terms = 0
    
    for word, tag in tagged:
        wn_tag = get_wordnet_pos(word)
        
        synsets = wordnet.synsets(word, pos=wn_tag)
        if not synsets:
            continue
        
        synset = synsets[0]
        
        try:
            swn_synset = swn.senti_synset(synset.name())
            sentiment_score += swn_synset.pos_score() - swn_synset.neg_score()
            valid_terms += 1
        except:
            continue
    
    if valid_terms == 0:
        return 1
    
    avg_score = sentiment_score / valid_terms
    return 1 if avg_score >= 0 else 0

df_models["pred_sentiwordnet"] = df_models["text"].apply(sentiwordnet_predict)

print(df_models[["text", "label", "pred_sentiwordnet"]].head())
```

Avaliação dos modelos léxicos

```python
def evaluate_model(y_true, y_pred, model_name, show_report=False):
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1-score": f1_score(y_true, y_pred)
    }
```

```python
# Avaliar cada modelo e armazenar os resultados
results_lexicon = []

results_lexicon.append(evaluate_model(df_models["label_num"], df_models["pred_textblob"], "TextBlob"))
results_lexicon.append(evaluate_model(df_models["label_num"], df_models["pred_vader"], "VADER"))
results_lexicon.append(evaluate_model(df_models["label_num"], df_models["pred_stanza"], "Stanza"))
results_lexicon.append(evaluate_model(df_models["label_num"], df_models["pred_sentiwordnet"], "SentiWordNet"))

results_lexicon_df = pd.DataFrame(results_lexicon)
print(results_lexicon_df)
```

```python
def create_unified_csv(pos_file, neg_file, output_filename):
    data = []

    # Processar palavras Positivas
    with open(pos_file, 'r', encoding='utf-8', errors='ignore') as f:
        for line in f:
            word = line.strip()
            if word and not word.startswith(';'):
                data.append([word.lower(), 1, 0])

    # Processar palavras Negativas
    with open(neg_file, 'r', encoding='utf-8', errors='ignore') as f:
        for line in f:
            word = line.strip()
            if word and not word.startswith(';'):
                data.append([word.lower(), 0, 1])

    # Converter para DataFrame e guardar como CSV
    df_lexicon = pd.DataFrame(data, columns=['English', 'Positive', 'Negative'])

    # Remover possíveis palavras duplicadas
    df_lexicon = df_lexicon.drop_duplicates(subset=['English'])

    df_lexicon.to_csv(output_filename, index=False)
    print(f"Ficheiro '{output_filename}' criado com {len(df_lexicon)} palavras.")
    
    return df_lexicon


df_lexicon = create_unified_csv(pos_file, neg_file, "opinion_lexicon.csv")
```

```python
# 6.c - Classificador SEM tratamento de negação (baseline)
def classify_lexicon_without_negation(text, lexicon):
    score_pos = 0
    score_neg = 0
    
    if isinstance(text, str):
        words = text.split()
    else:
        words = []
    
    # Agora iteramos sobre a lista de palavras gerada
    for w in words:
        if w in lexicon:
            score_pos += lexicon[w]['positive']
            score_neg += lexicon[w]['negative']
            
    # Determinar a classificação
    if score_pos > score_neg:
        classification = "pos"
    else:
        classification = "neg"
        
    return score_pos, score_neg, classification
```

```python
# 6.d - Classificador COM tratamento de negação 
NEG_WORDS = {
    "no", "not","rather", "couldn't", "wasn't", "didn't", "wouldn't", "shouldn't", "weren't", "don't", "doesn't", "haven't", 
    "hasn't", "won't", "wont", "hadn't", "never", "none", "nobody", "nothing", "neither", "nor", "nowhere", "isn't", "can't", 
    "cannot", "mustn't", "mightn't", "shan't", "without", "needn't", "hardly", "less", "little", "rarely", "scarcely", "seldom"
}

# Definir os prefixos e sufixos como tuplos
neg_prefixes = ('de', 'dis', 'il', 'im', 'in', 'ir', 'mis', 'non', 'un')
neg_suffixes = ('less',)

def classify_lexicon_with_negation(text, lexicon):
    score_pos = 0
    score_neg = 0
    
    if isinstance(text, str):
        words = text.split()
    else:
        words = []
        
    negation_affection = 0 # contador das palavras afetadas pela negação
    
    for w in words:
        
        # 1. Se for palavra de negação explícita, define as próximas 2 palavras como afetadas
        if w in NEG_WORDS:
            negation_affection = 2 
            continue 
            
        # 2. Verificar se a palavra tem os prefixos ou sufixos de negação
        has_negative_affix = w.startswith(neg_prefixes) or w.endswith(neg_suffixes)
        
        # 3. Se a palavra estiver no léxico
        if w in lexicon:
            val_pos = lexicon[w]['positive']
            val_neg = lexicon[w]['negative']
            
            # Se a palavra tem prefixo/sufixo de negação: inverte o sentimento e zera o contador
            if has_negative_affix:
                score_pos += val_neg
                score_neg += val_pos
                negation_affection = 0
                
            # Se não tem prefixo/sufixo, mas está sob o efeito das 3 palavras anteriores
            elif negation_affection > 0:
                score_pos += val_neg
                score_neg += val_pos
              
            
            # Caso normal (sem negação ativa e sem prefixo/sufixo)
            else:
                score_pos += val_pos
                score_neg += val_neg
                
        # 4. Se a palavra tem prefixo/sufixo MAS NÃO está no léxico
        # (Forçamos a classificação como negativa e zeramos o contador)
        elif has_negative_affix:
            score_neg += 1 
            negation_affection = 0

        # Reduz o contador de negação a cada palavra que passa
        if negation_affection > 0:
            negation_affection -= 1
            
    # Determinar a classificação final
    if score_pos > score_neg:
        return score_pos, score_neg, "pos"
    else:
        return score_pos, score_neg, "neg"
```

```python
import torch
import numpy as np
from transformers import (
    DistilBertTokenizerFast,
    DistilBertForSequenceClassification,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding  
)
from torch.utils.data import Dataset as TorchDataset
from sklearn.metrics import accuracy_score, f1_score

# ─ 1. VERIFICAR DISPOSITIVO 
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"A usar: {device.upper()}")

# ─ 2. DATASET COM TOKENIZAÇÃO 
class IMDBDataset(TorchDataset):
    """
    Lazy tokenization: tokeniza apenas quando o item é acedido,
    em vez de tokenizar tudo de uma vez no __init__.
    """
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts.tolist()
        self.labels = labels.tolist()
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=self.max_len
        )
        encoding['labels'] = self.labels[idx]
        return encoding

    def __len__(self):
        return len(self.labels)


# ─ 3. TOKENIZER E MODELO
tokenizer_ft = DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')
model_ft = DistilBertForSequenceClassification.from_pretrained(
    'distilbert-base-uncased',
    num_labels=2
)

data_collator = DataCollatorWithPadding(tokenizer=tokenizer_ft)

# ─ 4. DATASETS
train_dataset = IMDBDataset(df_train['text'], df_train['label_num'], tokenizer_ft, max_len=128)
test_dataset  = IMDBDataset(df_models['text'], df_models['label_num'], tokenizer_ft, max_len=128)

print(f"Treino: {len(train_dataset)} | Teste: {len(test_dataset)}")


# ─ 5. MÉTRICAS
def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    preds = np.argmax(predictions, axis=-1)
    return {
        'accuracy': accuracy_score(labels, preds),
        'f1': f1_score(labels, preds, average='binary')
    }


# ─ 6. TRAINING ARGUMENTS OPTIMIZADOS 
training_args = TrainingArguments(
    output_dir='./ft_results',
    num_train_epochs=2,
    per_device_train_batch_size=32,   
    per_device_eval_batch_size=64,   
    fp16=torch.cuda.is_available(),   
    dataloader_num_workers=4,         
    # Avaliação e logging
    eval_strategy='epoch',
    save_strategy='no',
    logging_steps=50,              
    report_to='none',
    # Optimizador
    learning_rate=2e-5,              
    warmup_ratio=0.1,                
    weight_decay=0.01,

    seed=42,
)

# ─ 7. TRAINER
trainer = Trainer(
    model=model_ft,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics,
    data_collator=data_collator,      
    processing_class=tokenizer_ft
)

print('A iniciar fine-tuning...')
trainer.train()
print('Fine-tuning concluído.')
```

```python
preds_ft = trainer.predict(test_dataset)
y_pred_ft = np.argmax(preds_ft.predictions, axis=1)

r = evaluate_model(y_test, y_pred_ft, 'ML: DistilBERT Fine-tuned')
all_results.append(r)
print(r)
```

```python
import requests
import json
import time

# ============================================================
# CONFIGURAÇÃO DA API — iaedu.pt
# ============================================================
API_URL    = os.getenv('IAEDU_API_URL', '')
API_KEY    = os.getenv('IAEDU_API_KEY', '')
CHANNEL_ID = os.getenv('IAEDU_CHANNEL_ID', '')
THREAD_ID  = os.getenv('IAEDU_THREAD_ID', '')

# Formato SSE próprio da iaedu:
# {"run_id": "...", "type": "start"|"token"|"done"|"error", "content": "..."}

def llm_call_raw(message_text, delay_after=2.0):
    """Fazer uma chamada à API e devolver o texto completo da resposta."""
    form_data = {
        'channel_id': (None, CHANNEL_ID),
        'thread_id':  (None, THREAD_ID),
        'user_info':  (None, '{}'),
        'message':    (None, message_text),
    }
    resp = requests.post(
        API_URL,
        headers={'x-api-key': API_KEY},
        files=form_data,
        stream=True,
        timeout=60
    )
    resp.raise_for_status()

    full_text = ''
    error_msg = None
    for raw_line in resp.iter_lines():
        if not raw_line:
            continue
        line = raw_line.decode('utf-8') if isinstance(raw_line, bytes) else raw_line
        try:
            chunk = json.loads(line)
            ctype   = chunk.get('type', '')
            content = chunk.get('content', '')
            if ctype == 'token':
                full_text += content
            elif ctype == 'done' and not full_text:
                # Alguns agentes enviam a resposta completa no 'done'
                full_text = content
            elif ctype == 'error':
                error_msg = content
        except (json.JSONDecodeError, KeyError):
            pass

    time.sleep(delay_after)   # respeitar rate limit

    if error_msg:
        raise RuntimeError(f'API error: {error_msg}')
    return full_text.strip()


def llm_predict(text, system_prompt, max_retries=4, base_delay=3.0):
    """Classificar sentimento via iaedu.pt.
    
    O system_prompt e a instrução de classificação são incluídos na mensagem.
    Retorna 1 (pos), 0 (neg) ou -1 (erro / resposta ambígua).
    """
    full_message = (
        f"{system_prompt}\n\n"
        f"Classify the sentiment of this movie review as positive or negative. "
        f'Reply ONLY with "pos" or "neg". No other text.\n\nReview: {text[:1000]}'
    )

    for attempt in range(max_retries):
        try:
            answer = llm_call_raw(full_message, delay_after=base_delay + attempt * 2)
            answer_lower = answer.lower()
            if 'pos' in answer_lower:
                return 1
            elif 'neg' in answer_lower:
                return 0
            else:
                return -1   # resposta ambígua
        except RuntimeError as e:
            # Rate limit ou erro da API — esperar mais
            wait = base_delay * (2 ** attempt)
            print(f'  [tentativa {attempt+1}] {e} — aguardar {wait:.0f}s')
            time.sleep(wait)
        except Exception as e:
            if attempt < max_retries - 1:
                time.sleep(base_delay)
            else:
                print(f'  [erro após {max_retries} tentativas]: {e}')
                return -1
    return -1

# ============================================================
# Definição dos 3 prompts
# ============================================================
PROMPT_GENERIC = 'You are a sentiment analysis assistant. Classify text as positive or negative.'

PROMPT_DOMAIN = """You are a sentiment analysis assistant specialized in movie reviews from IMDB.
Reviews can be positive (the reviewer liked the movie) or negative (the reviewer disliked the movie).
Consider the overall tone, specific praise or criticism, and the reviewer's recommendation."""

PROMPT_FEWSHOT = """You are a sentiment analysis assistant for IMDB movie reviews.

Here are some examples:
- "This movie was amazing, great acting and beautiful cinematography!" -> pos
- "Terrible film, waste of time, the plot made no sense at all." -> neg
- "A masterpiece of cinema, one of the best films I've ever seen." -> pos
- "Worst movie I have ever seen, absolutely horrible acting and script." -> neg
- "Loved every minute, the director did an outstanding job." -> pos
- "Boring and predictable, I fell asleep halfway through." -> neg"""

print('API iaedu.pt configurada.')

# Teste rápido de ligação
test_result = llm_predict("This movie was absolutely fantastic!", PROMPT_GENERIC)
print(f'Teste: {"OK → pos" if test_result == 1 else "OK → neg" if test_result == 0 else "ERRO (-1)"}')
```

```python
SAMPLE_SIZE = 200
df_sample = df_models.sample(n=SAMPLE_SIZE, random_state=42).copy()

prompts = [
    ('LLM: Genérico', PROMPT_GENERIC),
    ('LLM: Domínio', PROMPT_DOMAIN),
    ('LLM: Few-shot', PROMPT_FEWSHOT)
]

for prompt_name, prompt in prompts:
    preds = []
    for text in tqdm(df_sample['text'], desc=prompt_name):
        preds.append(llm_predict(text, prompt))
        time.sleep(0.5)

    col = f'pred_{prompt_name}'
    df_sample[col] = preds

    valid_mask = df_sample[col] >= 0
    valid = df_sample[valid_mask]
    n_errors = (~valid_mask).sum()

    print(f'\n{prompt_name}: {len(valid)} válidas, {n_errors} erros')
    if len(valid) > 0:
        r = evaluate_model(valid['label_num'], valid[col], prompt_name, show_report=True)
        all_results.append(r)
        print(r)
    print()
```
